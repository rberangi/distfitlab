# distfitlab

Fit probability distributions to your data and compare them, from one widget UI.

- **Continuous** — 18 distributions (Normal … Student-T), fitted with `scipy.stats`
- **Discrete** — 5 count distributions (Poisson, Binomial, Geometric, Negative Binomial, Zero-Inflated Poisson)

Use the switch at the top to choose the family. **Fit All** fits every distribution and
ranks them; **Fit** checks parameters you enter; the **Data view** tab plots the data itself.

Files you save land under your current working directory (`outputs/`, `artifacts/`).
